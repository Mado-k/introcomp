# **Operadores Condicionais**

Imagine que você deseja calcular automaticamente a sua média em uma matéria da escola e saber se está aprovado ou reprovado apenas inserindo suas notas.

Para que você consiga escrever um programa com essas funcionalidades é necessário o uso dos operadores condicionais `if` e `else`, que serão o foco dessa aula.

Com eles, você poderá escrever um código que resolve o seu problema:

````python
# Suponha que a média da sua escola seja 6 para ser aprovado

nota_1 = 7
nota_2 = 6
nota_3 = 8

# Execute o código e depois mude a sua nota_3 para 4 e veja o que acontece
# Não se preocupe com a parte abaixo, brinque com as notas e olhe as saídas

# Abaixo, o calculo para a média de três numeros
media = (nota_1 + nota_2 + nota_3) / 3

print(f"A sua média foi {media:.2f} e você...")

if media >= 6:
  print("Está aprovado!")
else:
  print("Vai ter que fazer prova final...")
````

Não se importe com a estrutura por enquanto, isso foi só um exemplo! Os operadores condicionais são a sua principal ferramenta em Python para realizar **desvios de fluxo** no seu programa. Esses desvios são nada mais nada menos que possibilidades de mudar o curso do seu código.

A sua **condição** no problema da média escolar era a nota ser maior do que 6... ou não! Dependendo da resposta, o código seguiu um caminho diferente e te exibiu outra mensagem: aprovado ou reprovado. As duas não acontecem ao mesmo tempo!

Ou seja, **condição** é um requisito que o seu código deve atender para que um **desvio de fluxo** seja acionado.

# **Operador if**

O primeiro operador condicional que temos é o `if`, que pode ser traduzido para `se`.

Sua principal função é garantir que determinadas instruções só aconteçam caso a condição atribuída ao `if` seja verdadeira.

```python
if 'Alguma condição':
    'Alguma ação'
```

A figura abaixo (fluxograma) ilustra o funcionamento esperado de um condicional `if` (Figura 3):<br>

*   **2. Teste de condição**: é o próprio `if` e sua condição.

*   **4. Demais comandos**: são os comandos que acontecerão após o `if`, independente do valor lógico dele. Essa opção pode representar também o fim do código, ou seja, não acontece nada após os passos anteriores.

Seguindo com os exemplos:

* Note que o print não será executado no bloco abaixo, pois `tenho_biscoito` tem valor lógico falso, ou seja, não satisfaz a condição do `if`!



````python
tenho_biscoito = False
if tenho_biscoito:
    print("Tenho sim, pode pegar um pouco!")
````

* Aqui o print será executado no bloco abaixo, pois `tenho_biscoito` tem valor lógico verdadeiro, satisfazendo a condição do `if`!

````python
tenho_biscoito = True
if tenho_biscoito:
    print("Tenho sim, pode pegar um pouco!")
````

Neste momento, você pode ter notado o espaçamento à esquerda do comando `print()` e se perguntado o porquê. Esse espaçamento é chamado de  **indentação** e é **indispensável** para que códigos em Python funcionem corretamente, ajudando você e a máquina a interpretarem o código sem erros, além de proporcionar uma formatação menos poluída e mais bonita!

Para indentar, basta teclar `TAB`, a tecla acima do `Caps-Lock` (ou `fixa`, em alguns modelos de teclados) no seu teclado e o seu editor de texto ou ambiente de desenvolvimento provavelmente deixará o espaço necessário.


Abaixo veremos um exemplo que exemplifica a importância da indentação.

````python
idade = 10

# Verifica se tem idade para ver o filme

if idade >= 16:
  print("Você tem mais de 16 anos!")
print("Você pode ver o filme com classificação +16!")
````

Como é possível observar, o fato do segundo print não estar indentado, faz com que a pessoa receba que pode ver o filme, mesmo não podendo. Sendo assim, é indispensável o uso de indentação para um funcionamento correto do código.

Agora, para praticar com outros tipos de dados, vamos comparar dois valores e imprimir na tela se o valor comparado for menor que o outro valor.

**AVISO**: rode este primeiro trecho abaixo antes dos outros para evitar possíveis erros!

````python
# Sinta-se livre para editar os valores das variáveis e ver diferentes resultados nos blocos a seguir
valor_a = 10
valor_b = 20
````

````python
print("Início do código!")

# 1. Teste de condição (se valor_a é menor que valor_b)
if valor_a < valor_b:
  print(valor_a)

print("Fim do código!")
````

````python
print("Início do código!")

# 2. Teste de condição (se valor_b é menor que valor_a)
if valor_b < valor_a:
  print(valor_b)

print("Fim do código!")
````

Vale ressaltar que, quando um `if` não tem seu valor lógico satisfeito (ou seja, verdadeiro), os comandos dentro dele não serão executados, mas todos os comandos posteriores serão.

Para exercitar, veja mais um exemplo um pouco mais robusto, envolvendo vários testes de condições ligados ao valor da mesada recebida por uma pessoa.

````python
mesada = 350

if mesada < 100:
    print("Parabéns, sua avó te deu mais R$50,00!")
    mesada = mesada + 50

if mesada >= 100 and mesada < 300:
    print("Vou juntar mais um pouquinho pro GTA VI!")
    mesada = mesada + 30

if mesada >= 300 and mesada < 800:
    print("Hoje eu vou comer muito sorvete e doce com minha namorada!")
    mesada = mesada - 300

if mesada >= 800:
    print("Vou gastar tudo em cartinha de Pokemon!")
    mesada = 0

print(mesada)
````

Note que, ao final do código, `print(mesada)` será executado independentemente de alguma condição ter sido ou não satisfeita, já que está fora da indentação de qualquer `if`!

### **If com Strings**

Há uma diversidade enorme de funções para Strings que te respondem com valores booleanos `True` e `False`. Algumas delas são `isupper()`, `islower()` e `isdecimal()`.

Ao utilizar cada uma dessas funções, é como se você estivesse fazendo uma pergunta pra sua String, por exemplo:

````python
minha_string = "INTROCOMP"
if minha_string.isupper():
  print("A string está em maiúsculo!")
````

No código acima, imagine que a função `isupper` estivesse perguntando "você está toda em maiúsculo?" para a String e a resposta foi `True` (Verdade), então o print foi acionado!

````python
minha_string = "introComp"
if minha_string.islower():
  print("A string está em minúsculo!")
````

Nesse último caso, a resposta foi `False` (Falso), pois a pergunta foi "você está toda em minúsculo?", mas existe uma letra em maiúsculo, o `C` em `introComp`. Assim, a **condição** não foi atendida e nenhuma mensagem foi exibida.

Altere o `C` para minúsculo (`c`) e veja a mudança na resposta!

Por último, `isdecimal()` te diz se a String é um número inteiro positivo:

````python
minha_string = "2025" # Tente também com outros valores como "numero", "-1" e "1.2"
if minha_string.isdecimal():
  print("A string é um número inteiro positivo!")
else:
  print("A string não é um número inteiro positivo!")

# Preste atenção: essa função também vai retornar falso para números negativos ou reais!!!
````

Caso tenha interesse, veja mais métodos de `String` na documentação de Python, disponível nesse [link](https://docs.python.org/pt-br/3/library/stdtypes.html#string-methods).

# **Operador else**

O `else` é um operador que pode ser definido quase como um "complementar" do `if`, sendo uma tradução para `se não` ou `caso contrário`. Ele é usado em situações onde mais de uma ação pode ocorrer, dependendo de determinada circunstância.









Por exemplo: Imagine que você está pensando em dar um passeio. Antes de sair, você resolve dar uma olhada em como está o tempo. SE estiver chovendo, você pegará seu guarda-chuva para não se molhar; SE NÃO, você pega seu óculos de sol.

Observe que, após analisar o clima, o passeio prossegue da mesma forma. A diferença estava apenas na decisão que você precisou tomar antes de sair de casa. O mesmo vai acontecer nos códigos, existem trechos que precisam seguir caminhos diferentes, mas depois se encontram e voltam a executar as mesmas instruções.

Como você viu nas imagens, o `else` é o que acontece do outro lado, ou seja, se a condição do seu `if` for `False`, então o código seguirá para os comandos do `else`.
Perceba que em ambos a indentação é extremamente importante para o funcionamento do código.

```python
if 'Alguma condição':
    'Alguma ação'
else:
    'Alguma outra ação, caso a condição não seja atendida'
```


E se você quiser imprimir o maior valor entre **a** e **b**? Utilizando o fluxograma da Figura 6 para visualizar o processo, tome as variáveis **a** e **b**, de valores desconhecidos. O teste de condição será: `a > b` .

Caso essa afirmação seja verdadeira, ou seja a > b, a instrução executada será: imprima o valor de **a**. Do contrário, a afirmação é falsa e uma outra instrução é executada: imprima o valor de **b**.

E lembrando da nossa aula 3, podemos utilizar várias operações lógicas dentro de uma condição! Imagine que você queira jogar com os seus amigos Sophie, Victor e Marco, mas precisa verificar se todo mundo tem disponibilidade. Cada um avisa qual horário pode jogar e você precisa verificar se é compatível com o seu. Para alguém ser considerado disponível, basta aquela pessoa estar livre na mesma hora que você.

Observe o código abaixo:

````python
# Mude os horarios para verificar o que acontece. Cada variavel representa o horario que aquela pessoa pode jogar.
meu_horario = 20
horario_sophie = 20
horario_marco = 21
horario_victor = 19

# Existem 3 condições no if e cada uma dessas condições são operações lógicas que retornam "True" ou "False"
# Se o horário de um amigo for igual ao meu_horario, então a condição é "True", caso contrário, "False"
# Se todo mundo estiver disponivel, vamos poder jogar. Caso contrario, nao vamos poder jogar :(
if (horario_sophie == meu_horario) and (horario_marco == meu_horario) and (horario_victor == meu_horario):
    print("Vamos poder jogar!")
else:
    print("Não vamos poder jogar :(")
````
