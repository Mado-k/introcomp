## **Tipos de Dados**

Você já conhece os quatro tipos primitivos fundamentais do Python. Relembre o que cada um representa:

**`int` (inteiro):** Números inteiros, positivos ou negativos, sem casas decimais. Exemplos: idade de uma pessoa, quantidade de itens, número de páginas de um livro.

**`float` (ponto flutuante):** Números com casas decimais, usados para medidas precisas. Exemplos: altura, peso, temperatura, preço de produtos.

**`bool` (booleano):** Valores lógicos que representam apenas duas possibilidades: `True` (verdadeiro) ou `False` (falso). Exemplos: aprovado/reprovado, ligado/desligado, presente/ausente.

* Seu saldo é suficiente

Após isso, seu código deve imprimir `Gasolina comprada, agora seu orçamento é de [valor]`.


**Casos de Teste**


| Entrada                | Saída                                                                                      |
| ---------------------- | ------------------------------------------------------------------------------------------ |

Imagine que você está programando a lógica principal de pontuação para um pequeno jogo 2D. O seu personagem começa a fase com 0 moedas.

Crie um programa que pergunte repetidamente ao jogador: "Você pegou uma moeda? (s/n): ".

Enquanto a resposta for **"s"**, o programa deve adicionar **10 pontos** à pontuação total do jogador e exibir a mensagem: **"Mais 10 pontos! Pontuação atual: X"**.

Se o jogador digitar **"n"**, ou qualquer outra coisa, o ciclo deve ser quebrado.

Ao final, quando o loop terminar, o programa deve exibir: **"Fim de jogo! Sua pontuação final foi de X pontos."**.

**Caso de teste**:
| Entrada | Saída Esperada |
| :--- | :--- |
| s<br>s<br>s<br>n | Mais 10 pontos! Pontuação atual: 10<br>Mais 10 pontos! Pontuação atual: 20<br>Mais 10 pontos! Pontuação atual: 30<br>Fim de jogo! Sua pontuação final foi de 30 pontos. |

````python
resp = str(input('Você pegou uma moeda? (s/n):'))
pts = 0
while resp.lower() == 's':
  pts += 10
  resp = str(input('Você pegou uma moeda? (s/n):'))
if pts < 100:
  print(f'Fim de jogo! Sua pontuação final foi de {pts} pontos.')
else:
  print(f'Fim de jogo! Sua pontuação final foi de {pts} pontos.\n E você é competitivo! Parabéns!')
````

### **Exercício 10**

Crie um pequeno simulador de processamento de dados de um sensor de temperatura. O programa deve solicitar repetidamente que o usuário insira uma "leitura do sensor" (um número inteiro). O loop deve rodar continuamente enquanto o usuário não digitar um valor negativo. O valor negativo servirá apenas como um "botão de parada" e não deve ser contabilizado nos cálculos.

Além disso, caso o valor da temperatura lida seja maior que **60°C**, o programa deve iniciar um processo de resfriamento, diminuindo a temperatura até que atinja **30°C** ou menos. Para isso, o programa deve exibir na tela a temperatura diminuindo de **5°C** em **5°C**. Os valores do resfriamento não devem ser contados no cálculo final.

Quando o usuário digitar um número negativo e o loop for encerrado, o programa deve imprimir a quantidade de leituras válidas realizadas e a temperatura média. Caso a primeira coisa que o usuário digite seja um número negativo, o programa deve exibir o aviso: "Nenhum dado válido foi inserido para calcular a média."

**Casos de teste**:
| Entrada        | Saída Esperada                                         |
| :---           | :---                                                   |
| -1             | --- Relatório Final ---<br>Nenhum dado válido foi inserido para calcular a média. |
| 40<br>50<br>-5 | --- Relatório Final ---<br>Total de leituras válidas: 2<br>Temperatura média: 45.00°C|
| 65<br>-1       | Iniciando processo de resfriamento...<br>60°C<br>55°C<br>50°C<br>45°C<br>40°C<br>35°C<br>30°C<br><br>--- Relatório Final ---<br>Total de leituras válidas: 2<br>Temperatura média: 47.50°C |

````python
# Desenvolva aqui o sensor!
leitura = 0
l1 = leitura
leitura_total = 0
i = 0

while True:
  leitura = float(input())
  if leitura < 0 and i == 0:
    print('--- Relatório Final ---\n Nenhum dado válido foi inserido para calcular a média')
    break
  elif leitura > 0:
    leitura_total += leitura
    i += 1
  elif leitura < 0 and i > 1:
    temp_media = leitura_total / i
    print(f'--- Relatório Final ---\nTotal de leituras válidas: {i}\nTemperatura média: {temp_media:.2f}°C')
    break

  elif leitura > 60:
    print('Iniciando processo de resfriamento...')
    while leitura > 30:
      leitura -= 5
      print(f'{leitura}°C')
````

### **Exercício 7**

````python
# Entradas
distancia = float(input("Digite a distância: "))
consumo = float(input("Digite o consumo: "))
preco = float(input("Digite o preço: "))
orcamento_gasolina = float(input("Digite o orçamento: "))

# Cálculos necessários
litros_gastos = distancia/consumo
valor_gasto = litros_gastos * preco

print(f'Gastos com gasolina: {valor_gasto}')

if valor_gasto > orcamento_gasolina:
  print('A viagem não será possível')
else:
  # Note uma entrada que só ocorre sob essa condicão
  orcamento_geral = float(input("Digite o orçamento geral: "))
  sobra = orcamento_gasolina - valor_gasto
  orcamento_geral = orcamento_geral + sobra

  print(f'Gasolina comprada, agora seu orçamento é de {orcamento_geral:.2f}')
````

### **Exercício 8**

````python
pontuacao = 0
continuar = "s"

print("Bem-vindo à fase de coleta de moedas!")

# O loop roda enquanto a variável 'continuar' for exatamente igual a "s"
while continuar == "s":
    continuar = input("Você pegou uma moeda? (s/n): ")

    if continuar == "s":
        pontuacao += 10
        print(f"Mais 10 pontos! Pontuação atual: {pontuacao}")

print(f"\nFim de jogo! Sua pontuação final foi de {pontuacao} pontos.")
````

### **Exercício 9**

````python
numero = int(input("Digite um número inteiro positivo: "))
print(f"Os divisores de {numero} são: ")


# O loop vai testar todos os números de 2 até o próprio (número digitado - 1)
i = 2
while(i < numero):
    # Se o resto da divisão for zero, 'i' é um divisor válido
    if numero % i == 0:
        print(i)
    i+=1
````

###**Exercício 10**

````python
soma_temperaturas = 0
quantidade_leituras = 0

print("Insira as leituras do sensor. Digite um valor negativo para encerrar.")

while True:
    leitura = int(input("Leitura do sensor: "))

    # Condição de parada imediata
    if leitura < 0:
        break

    # Contabiliza o valor da temperatura atual
    soma_temperaturas += leitura
    quantidade_leituras += 1

    # Processo de resfriamento
    if leitura > 60:
        print("\nIniciando processo de resfriamento...")
        while leitura > 30:
            leitura -= 5
            print(f"{leitura}°C")

print("\n--- Relatório Final ---")
if quantidade_leituras > 0:
    media = soma_temperaturas / quantidade_leituras
    print(f"Total de leituras válidas: {quantidade_leituras}")
    print(f"Temperatura média: {media:.2f}°C")
else:
    print("Nenhum dado válido foi inserido para calcular a média.")
````
