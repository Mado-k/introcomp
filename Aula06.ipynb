# **While**

O comando `while` (que significa "enquanto", em inglês) é uma estrutura de repetição que executa um bloco de código de forma contínua **enquanto uma determinada condição for verdadeira.**

Ele é a ferramenta ideal para situações em que você **não sabe** de antemão quantas vezes o loop precisará se repetir. O programa simplesmente continuará executando aquela tarefa até que a condição seja quebrada.

Veja como criar um loop `while` em Python:

```python
while <condicao>:
  <comando_1>
  <comando_2>
  <comando_3>
  ...
```

**IMPORTANTE**!
Assim como você já fazia nos comandos `if`, preste muita atenção aos detalhes da sintaxe do Python: não se esqueça dos dois pontos (`:`) após a condição e lembre-se da indentação (o recuo do texto) nos comandos internos.

## **Inicialização**

Antes de abrirmos a porta do loop, precisamos criar a nossa **variável de controle.** É o valor dessa variável que ditará quando a repetição deve chegar ao fim.

Para começar a entender, vamos criar o caso mais simples possível: um contador numérico. *(Mas guarde essa informação: a variável de controle também poderia ser uma palavra, um texto digitado pelo usuário ou qualquer outra condição!)*.

````python
contador = 1
````

## **Condição**

A condição é a grande "tomadora de decisão" do nosso loop: é ela que vai definir se o código continuará sendo executado ou se será interrompido.

Ainda usando o nosso exemplo do contador, nós podemos definir um "teto", ou seja, o valor máximo que o nosso loop aceita rodar:

```python
while contador <= 5:
  print(contador)
```

Neste exemplo, enquanto o valor da variável `contador` for menor ou igual a 5, o bloco de código interno (o `print`) continuará sendo executado.

Mas olhe atentamente para o código acima... *você já consegue ver algum problema grave no que construímos até o momento?*

## **Atualização**

Se você percebeu que o valor do contador nunca muda, você está certíssimo!

Como não dissemos ao código que ele deveria adicionar números ao contador, ele permanecerá valendo 1 para sempre. Consequentemente, a condição contador <= 5 nunca deixará de ser verdadeira, criando o que chamamos de loop infinito.

Para solucionar isso, precisamos atualizar o valor da nossa variável de controle a cada iteração (cada volta do loop):

````python
while contador <= 5:
  print(contador)
  contador += 1
````

# **Operações com While**

## **Contadores**

Você viu que o while repete uma ação enquanto uma condição for verdadeira. Mas aqui surge um problema: o computador não tem dedos para contar! Se você disser para ele "dê 3 pulos", ele não sabe sozinho quantos pulos já deu. Precisamos criar uma "memória" para ele acompanhar isso. É aí que entra o Contador.

Um contador é simplesmente uma variável que usamos para registrar quantas vezes o nosso laço (loop) já rodou.

Para um contador funcionar perfeitamente em um while, ele precisa de 3 passos obrigatórios:

**Nascer (Inicialização)**: Criamos a variável antes do loop começar, geralmente valendo zero.

**Viver (A Condição)**: O while verifica se o contador já chegou no limite que queremos.

**Crescer (A Atualização)**: Dentro do loop, nós atualizamos o valor do contador.

### **Exemplo prático**

````python
# 1. NASCER: Criamos o contador valendo 0
contador = 0

# 2. VIVER: O loop continua ENQUANTO o contador for menor que 3
while contador < 3:
    print("Dei um pulo!")

    # 3. CRESCER: A ATUALIZAÇÃO! (Isso é crucial)
    # Pegamos o valor antigo, somamos 1, e guardamos na mesma variável
    contador = contador + 1

print("Terminei de pular!")
````

## **Acumuladores**

No passo anterior, vimos que o contador serve para registrar quantas vezes algo aconteceu. Ele sempre cresce de forma fixa (geralmente de 1 em 1).

Mas e se precisarmos somar valores que mudam o tempo todo? É aqui que entram os **Acumuladores**.

Um acumulador é como um cofrinho de moedas.


*   Se você colocar 3 moedas no cofrinho, um contador diria: "Você colocou 3 itens aqui".
*   Já o acumulador diria: "Você colocou uma moeda de R\$ 1,00, uma de R\$ 0,50 e uma de R\$ 0,25. O valor total acumulado é R\$ 1,75".

Enquanto o contador soma um valor constante (contador = contador + 1), o acumulador soma um valor variável (acumulador = acumulador + novo_valor).

### **Exemplo prático**

Vamos imaginar que estamos escrevendo o sistema do caixa da cantina. O programa vai perguntar o preço de 3 lanches e, no final, dizer o total que o aluno precisa pagar.

Neste exemplo, vamos usar um contador (para controlar os 3 lanches) e um acumulador (para somar o dinheiro) trabalhando juntos:

````python
# 1. INICIALIZAÇÃO
lanches_passados = 0   # Este é o nosso CONTADOR (quantos lanches já passaram)
total_a_pagar = 0.0    # Este é o nosso ACUMULADOR (o valor em dinheiro)

# 2. CONDIÇÃO
while lanches_passados < 3:
    # Pedimos o preço do lanche atual
    preco = float(input("Digite o preço do lanche: R$ "))

    # 3. ATUALIZAÇÃO DO ACUMULADOR:
    # O novo total é o que já tínhamos + o preço do novo lanche
    total_a_pagar += preco

    # 3. ATUALIZAÇÃO DO CONTADOR:
    lanches_passados += 1

print(f"Fim da compra! O total a pagar é R$ {total_a_pagar}")
````

## **Validação de Dados**

Até agora, você viu o while sendo usado para contar ou acumular valores. Mas existe uma terceira utilidade extremamente comum na programação: usar o loop para **proteger o seu programa de usuários distraídos** (ou mal-intencionados). Isso se chama **Validação de Dados**.

Imagine que o seu programa faz uma pergunta simples de "Sim" ou "Não" e pede para o usuário digitar S ou N. O que acontece se ele digitar Batata? O programa pode quebrar ou fazer um cálculo completamente errado.

É aqui que o while cumpre o papel de segurança. Ele fica barrando a entrada do usuário e dizendo: "Enquanto você não digitar o que eu pedi, você não passa para a próxima linha do código".

### **A lógica**

Para validar dados, a lógica da condição do while muda um pouquinho. Nós não vamos repetir enquanto algo estiver "certo", mas sim enquanto a resposta estiver ERRADA.

A estrutura funciona nestes 3 passos:


1.   **Pedimos a informação** a primeira vez.
2.   **Avaliamos o erro**: O while verifica se o que o usuário digitou é inválido.
3.   **Insistimos:** Dentro do loop, mostramos uma mensagem de erro e pedimos a informação de novo.









### **Exemplo prático**

Imagine a tela final de um jogo ou o menu de um aplicativo perguntando: "Deseja salvar e sair?". Para o programa funcionar, ele espera apenas duas respostas possíveis: S (Sim) ou N (Não). Se o usuário esbarrar no teclado e digitar "X", o computador não vai saber o que fazer.
Veja como esse código clássico de "Sim ou Não" é simples e poderoso na prática:


````python
resposta = input("Deseja continuar? (S/N): ")

# O loop trava o usuário ENQUANTO ele não digitar 'S' nem 'N'
while resposta != "S" and resposta != "N":
    print("Erro! Por favor, digite apenas S ou N.")

    # Pedimos de novo para atualizar a variável e não criar um loop infinito
    resposta = input("Deseja continuar? (S/N): ")

print("Opção válida recebida! O programa pode seguir.")
````

# **While True**

Em certas situações, apesar de incialmente contraintuitivo, é interessante construir um loop **sem condição de parada**. Para isso, utilizaremos o valor booleano `True` como condição para o `while`.

````python
while True
````

Utilizamos esse tipo de loop em situações específicas, comumente quando um programa depende de entradas constantes do usuário, ou quando é necessário executar o loop **pelo menos uma vez** (visto que `while True` sempre será executado quando o programa chegar nele).

## **Break**

A expressão de condição do `while` ser simplesmente `True` implica que a estrutura nunca irá encontrar uma condição de parada. Para isso, linguagens de programação implementam ferramentas que encerram loops (chamadas de *goto restritos*), mesmo que sua condição não tenha se tornado `False`. Em python, o comando `break` (quebrar, em inglês) é utilizado exclusivamente dentro de estruturas de repetição para **encerrá-las** imediatamente.

 Como um loop *literalmente* infinito é computacionalmente indesejado, operações `while True` utilizam o `break` para terminar o laço, normalmente executados dentro de estruturas condicionais `if/elif/else`.

````python
while True:

  # Código a ser repetido
  # ...

  if condicao:
    break # O laço para aqui
````

O break também é utilizado quando se quer encerrar um loop **prematuramente**, sem que todas as linhas da estrutura sejam executadas antes do laço checar novamente a condição de parada e finalmente encerrar. Quando o `break` é executado, o loop é encerrado exatamente na linha do break, ocorrendo um salto daquela região para o resto do código fora da estrutura.

## **Exemplo**

Em interfaces que implementam menus, onde há várias opções e comandos a serem escolhidos, é interessante que o programa rode em loops extensos, o qual terminam quando o usuário clique em **sair** ou **avançem** em algum processo específico. Em jogos eletrônicos, é muito comum esse tipo de implementação, observe o código abaixo:

````python
while True:

  comando = input("Digite um comando: ")

  if comando == 'continuar':
    print("Jogo retomado a partir do último save!")

  if comando == 'novo jogo':
    print("Novo jogo criado!")

  if comando == 'salvar':
    print("Jogo salvo!")

  if comando == 'sair':
    print("Fechando o jogo.")
    break # Loop é encerrado aqui

print("Obrigado por jogar!")
````

Analisando esse código, incialmente pode ser levantada a dúvida: e se fosse simplesmente atribuído a condição do loop como `comando != 'sair'`? Para explicar como isso pode não ser interessante, tente copiar o código abaixo no VS Code (ou no editor de texto de sua preferência) e rodá-lo (no colab, o erro não será exibido caso a última célula tenha sido executada).

````python
# CÓDIGO COM ERRO! Execute-o no VS Code
while comando != 'sair':

  comando = input("Digite um comando: ")
  if comando == 'continuar':
    print("Jogo retomado a partir do último save!")

  if comando == 'novo jogo':
    print("Novo jogo criado!")
````

O erro exibido indica que a váriavel `comando` ainda não foi definida. Isso ocorre porque ela foi declarada **dentro** do while, ou seja, quando o programa chegar em `while comando != sair`, ele irá procurar uma denifição de `comando`, que ainda não existe. Utilizando `while True`, é garantido que o loop irá rodar **pelo menos uma vez**, portanto a variável `comando` será definida inevitavelmente, contornando o problema.

Um jeito de escrever o código do exemplo acima, sem a utilização de breaks, seria:

````python
# Código sem break

comando = input("Digite um comando: ")

while comando != 'sair':

  if comando == 'continuar':
    print("Jogo retomado a partir do último save!")

  if comando == 'novo jogo':
    print("Novo jogo criado!")

  if comando == 'salvar':
    print("Jogo salvo!")

  comando = input("Digite um comando: ")

print("Fechando o jogo.")
print("Obrigado por jogar!")
````

Perceba como isso cria **repetição** e **redundância** na escrita do código, sendo necessário reescrever linhas, o que não é ideal, para que um resultado satisfatório seja obtido.

O break é uma ferramenta bastante conveniente na confecção de programas em python (e em muitas outras linguagens de programação!).

Faça um programa que receba como entrada opções, representadas por palavras, e, a partir delas, faça tarefas específicas. As opções são:

* Opção "adicionar": recebe dois números da entrada e imprime a soma resultante entre eles.


* Opção "multiplicar": recebe dois números da entrada e imprime o produto da multiplicação entre eles.


* Opção "repetir": recebe uma string e um número, imprime a string o número de vezes que esse número indicar. (dica: multiplicação entre strings e inteiros)

* Opção "sair": encerra o programa.

* Caso a entrada não for nenhuma das palavras acima, imprima "Comando inválido.".

````python
action = input('O que deseja realizar? (adicionar, mult, repeat, sair)')

while True:
  if action == 'adicionar':
    num1 = float(input())
    num2 = float(input())
    soma = num1 + num2
    print(soma)
    action = input()

  elif action == 'mult':
    num1 = float(input())
    num2 = float(input())
    mult = num1 * num2
    print(mult)
    action = input()

  elif action == 'repeat':
    action = input('(adicionar, mult, repeat, sair): ')
    action = input()

  elif action == 'sair':
    break

  while action != 'adicionar' and action != 'mult' and action != 'repeat' and action != 'sair':
    print('Você precisa inserir um valor válido!')
    action = input('O que deseja realizar? (adicionar, mult, repeat, sair)')

print('Boa utilização!')
````

# **Erros Comuns**

É muito importante dobrar a atenção ao trabalhar com loops, pois existem alguns erros "bobos" que podem acontecer e custar caro. Dentro desses, se destacam:

- Esquecer de mudar uma variável cuja condição de parada depende dela (*Loop infinito indesejado*);
- Esquecer a própria condição de parada;
- Condição falsa desde o princípio;
- Esquecer de inicializar uma variável contadora ou booleana antes de usá-la no loop.

Esses erros podem te levar a problemas simples como a não execução de um laço até erros mais complexos como o chamado loop infinito.

Conheça alguns dos comportamentos relacionados a tais erros nos exemplos abaixo:

Quando o código não tiver uma condição de parada ou você esquecer de mudar o seu estado, o programa poderá entrar em um loop infinito. Veja na próxima célula de código um exemplo, mas **cuidado** ao executá-la! A execução de qualquer programa pode ser interrompida com o comando `Ctrl + C` no terminal, então lembre-se de usá-lo caso isso aconteça.

Para interromper a célula abaixo, é preciso que você clique no círculo com o quadrado no meio duas vezes. Irá aparecer uma mensagem sobre como estará interrompendo a célula.

````python
while True:
    print("Ah não, estou preso num loop infinito!")
````

No programa, a mensagem "Ah não, estou preso num loop infinito!" não deixava de aparecer. Isso ocorreu porque no `while True` não houve implementação de `break` ou uma condição de parada.

````python
numero = 42

while numero < 5 :
    print("O numero é menor do que 5")
    numero += 1
````

O código dentro desse laço nunca será executado. Por que?

É bom estar atento à condição do `while`. O número inicia com o valor 42, porém o while só é executado quando o número é menor que 5. Dessa forma, as instruções dentro do while são ignoradas desde o começo.

````python
while contador <= 10:
    print(contador)
    contador += 1
````

Observe como é o mesmo erro obtido na sessão **while True**. Esse erro está sinalizando que a variável de nome `contador` não foi definida. Em Python, se você quer verificar o valor de uma variável, comparar a um outro valor, somar com outro, etc., essa variável precisa inicialmente ser definida e atribuída um valor.

Isso é resolvido se antes do `while`, um valor inicial for atribuído à variável `contador` (ou utilizando um `while True` como visto anteriormente).

**IMPORTANTE:**
* Todos os erros mencionados são desatenções recorrentes para programadores de todos os níveis. Portanto, é muito importante que, nessas situações, você seja capaz de *diagnosticar* a origem do problema. A prática torna esse processo cada vez mais natural, então não tenha medo de errar!

# **Resumo da Aula**



*   **Loops (Laços de Repetição):** São estruturas que permitem automatizar a execução de um bloco de código várias vezes, evitando repetições manuais e longas.
*   **Comando ``while``:** Repete uma ação enquanto uma condição for verdadeira. Exige sempre a regra de ouro em três passos: Inicialização, Condição e Atualização.
*   **Contadores e Acumuladores:** O contador soma um valor fixo a cada volta (útil para saber quantas vezes o loop rodou). O acumulador soma valores variáveis (útil para guardar um total acumulado, como um cofrinho).
*   **Validação de Dados:** O while pode atuar como um "segurança", repetindo uma pergunta até que o usuário insira uma resposta válida para o programa continuar.
*   **``while True`` e ``break``:** Usados em conjunto para criar laços contínuos que só são encerrados "de dentro para fora" quando o comando break é acionado.
*   **Cuidado com os Erros!** Lembre-se sempre de inicializar suas variáveis antes do laço e de garantir que a variável de controle seja atualizada, ou você poderá criar um loop infinito indesejado.

## **Exercício 1**

Existem algumas soluções possíveis. A primeira é a "mais simples", mas ambas estão corretas

````python
# Inicialização do contador
contador = 2

# Loop roda enquanto contador não atinge o máximo
while contador <= 100:
  print(contador)
  contador += 2 # Atualização do contador para o próximo número par
````

````python
# Inicialização do contador
contador = 1

# Loop roda enquanto contador não atinge o máximo
while contador <= 100:
  # Imprime apenas os números pares
  if contador % 2 == 0:
    print(contador)
  # Atualiza o contador para o próximo número
  contador += 1
````

Além da diferença no código, note que o primeiro irá realizar passos maiores, precisando de um número menor de passos totais. Isso faz com o que o primeiro código leve menos tempo para ser executado.

## **Exercício 2**

````python
# Leitura do valor de parada
valor = int(input("Digite um número: "))

# Inicialização das variáveis
contador = 1
soma = 0

while contador <= valor:
  soma += contador
  contador += 1

print(f"A soma dos números de 1 até {valor} é igual a {soma}")
````

## **Exercício 3**

````python
# Inicialização das variáveis
total_pontos = 0
quantidade_partidas = 0

print("--- Registro do Torneio ---")
print("Digite -1 a qualquer momento para encerrar.\n")

# Lemos a primeira pontuação antes de entrar no loop
pontos = int(input("Digite a pontuação da partida: "))

# O loop roda ENQUANTO o valor digitado não for negativo
while pontos >= 0:
    # Crescer: atualizamos o acumulador e o contador
    total_pontos += pontos
    quantidade_partidas += 1

    # Pedimos a próxima pontuação no final do loop
    pontos = int(input("Digite a pontuação da próxima partida: "))

# Resultados finais
print("\n--- RESUMO DO JOGADOR ---")
print(f"Partidas jogadas: {quantidade_partidas}")
print(f"Pontuação total: {total_pontos}")
````
