# **Lógica Relacional**

As **comparações** que relacionam valores e variáveis são denominadas
**relações lógicas**, sendo avaliadas como **True** (verdadeiro) ou **False** (falso). Esses são os dois valores possíveis de variáveis de tipo *bool*, que são usadas para guardar os resultados de uma relação lógica.

Em Python, essas comparações são feitas por meio de operadores. São eles: `>, <, >=, <=, ==, !=`. Para entender o que eles significam, observe a tabela:

| Operação | Significado |
| :--- | :--- |
| **`A > B`** | **A maior que B** |
| **`A < B`** | **A menor que B** |
| **`A >= B`** | **A maior ou igual a B** |
| **`A <= B`** | **A menor ou igual a B** |
| **`A == B`** | **A igual a B** |
| **`A != B`** | **A diferente de B** |

Observe, a seguir, como usar em código os operadores apresentados.

### **Igual (==)**

O exemplo abaixo usa uma variável booleana para guardar o valor de uma relação lógica e depois imprime o resultado na tela:

````python
meu_booleano = (10 == 20) # Comparação: 10 é igual a 20?
print(meu_booleano) # Função para imprimir o valor na tela
````

Você já viu que o operador `=` realiza uma atribuição; portanto, ele não pode ser usado para verificar igualdade entre duas variáveis. Para resolver isso, foi adicionando um `=` ao operador, resultando em `==`. Observe o exemplo:

````python
# Verificando se 2 + 2 resulta em 4

resultado_conta = 2 + 2 # Aqui fazemos uma atribuição

comparacao = (resultado_conta == 4) # Aqui fazemos uma comparação

print(comparacao) # Impressão do valor booleano obtido acima
````

O código acima pode ser simplificado:

````python
# Você pode imprimir o resultado da comparação sem
# precisar guardá-lo numa variável
print((2 + 2) == 4) # Isso é bem legal!
````

Você também pode, por exemplo, verificar se uma variável é igual a um valor específico:

````python
numero_magico = 7.6 # Atribuição de valor à variável

print("Digite o número mágico: ")
palpite = float(input()) # Recebendo um número de entrada
comparacao = (numero_magico == palpite) # Comparando o número de entrada com o número mágico

print(f"{palpite} é o número mágico? {comparacao}") # Escrevendo o palpite e o resultado da comparação na saída

# Um print equivalente sem usar a variável comparacao:
# print(f"{palpite} é o número mágico? {numero_magico == palpite}")
````

Comparações também podem ser feitas em _strings_ e outros tipos não-numéricos. Veja a seguir:

## **Comparações entre strings**

Se as duas variáveis analisadas são do tipo *string*, um valor numérico é associado a cada letra, de acordo com a ordem alfabética (ou seja, o valor de "a" é menor que o valor de "b", que por sua vez tem um valor menor que o de "c", e por aí vai). Mas isso não é tudo: comparando a mesma letra em suas versões maiúscula e minúscula, elas terão o mesmo valor?

````python
# Serão consideradas as variáveis minúscula e maiúscula
minuscula = 'a'
maiuscula = 'A'

# Será que elas são iguais?
print(minuscula == maiuscula)
````

Isso ocorre porque, para o computador, elas são símbolos diferentes, uma vez que a máquina deve saber quando você quer utilizar uma letra minúscula ou uma maiúscula. Portanto, `'a'` e `'A'` não podem ter o mesmo valor, mesmo representando a mesma letra.

> **Interessante saber:**  No Python, os caracteres são codificados com base em um padrão, conhecido como tabela ASCII. Nessa tabela, cada caractere é representado por um código numérico. No caso das letras:
>
> * *Maiúsculas (A-Z):* códigos 65 a 90.
> * *Minúsculas (a-z):* códigos 97 a 122.
>
> Dessa forma, ao comparar `'A'`, que vale 65, com `'a'`, que vale 97, o resultado é `False`.

Veja mais exemplos de comparações entre strings:

````python
print("Joãozinho" == "joãozinho")
````

````python
print("Joãozinho" == "Joãozinho")
````

# **Álgebra Booleana**

Agora que você conhece as relações lógicas e o tipo booleano, está pronto para entender **álgebra booleana**, criada pelo matemático britânico George Boole.\
Na computação, praticamente todos os programas fazem uso da álgebra booleana. Ela introduz conceitos como operadores, valores e expressões que permitem modelar diversos problemas.

Por exemplo, imagine um cadeado que, para abrir, exige uma senha de três dígitos. Representando os dígitos inseridos como $D1$, $D2$ e $D3$, a senha só é correta se:


*   $D1$ for igual ao primeiro dígito da senha,
*   $D2$ for igual ao segundo,
*   $D3$ for igual ao terceiro.

\
Dessa forma, usando a álgebra booleana, para que a senha inserida seja aceita, as 3 condições precisam ser satisfeitas simultaneamente:

$Senha_{correta} = {D1}_{correto} \ E \ {D2}_{correto} \ E \ {D3}_{correto}$

ou, em Python:

```python
senha_correta = D1_correto and D2_correto and D3_correto
```

A álgebra booleana define três operadores básicos: `and`, `or` e `not`. As próximas seções irão se debruçar sobre eles.

### **Operador 'and'**

Traduzindo ao pé da letra, o nome do operador é "E" (exemplo: irei comer arroz **E** feijão). Ele é utilizado em casos onde uma afirmação só é verdadeira quando **todas** as especificações são cumpridas.

Para exemplificar uma situação assim, imagine que um professor dará nota 10 para os alunos que fizerem a atividade 4 **E** a atividade 6 de uma lista. Se o aluno fizer as duas, ótimo! Se fizer apenas alguma das duas, isso não é suficiente. Se não fizer nenhuma das duas solicitadas, ele estará em apuros!\
Com o operador `and` é assim: só está tudo certo se todas as condições forem atendidas! Em álgebra booleana, essa situação pode ser representada assim:

\
$Nota10= {Atividade4}_{feita} \ E \ {Atividade6}_{feita}$


Abaixo, há uma representação dessa situação por meio de uma *checklist interativa*: marcar a caixinha de cada atividade significa que ela foi concluída. Execute a célula, clique nas duas primeiras caixinhas e observe o efeito disso na última!  

````python
# @title O aluno tirou 10?
from ipywidgets import Checkbox, VBox, HTML

# criar widgets
atvd4 = Checkbox(description='Atividade 4')
atvd6 = Checkbox(description='Atividade 6')
resultado = HTML(value="<span style='color: red; font-weight: bold;'>❌ Nota 10: Não</span>")

# função para atualizar o resultado
def _atualiza(change):
    if atvd4.value and atvd6.value:
        resultado.value = "<span style='color: green; font-weight: bold;'>✅ Nota 10: Sim!</span>"
    else:
        resultado.value = "<span style='color: red; font-weight: bold;'>❌ Nota 10: Não</span>"

# observar mudanças em Atividade 4 e Atividade 6
atvd4.observe(_atualiza, 'value')
atvd6.observe(_atualiza, 'value')

# exibir os widgets
VBox([atvd4, atvd6, resultado])
````

Por fim, é possível usar Python para ilustrar esse cenário:

````python
# O 'True' nas duas variáveis indica que o aluno concluiu as duas atividades

atividade4_feita = True
atividade6_feita = True

nota_10 = (atividade4_feita and atividade6_feita)

print(f"O aluno tirou 10? {nota_10}")
````

#### **Negando Operações**

Suponha que uma prova seja composta por quatro questões: Q1, Q2, Q3 e Q4. Cada questão é considerada verdadeira (True) se foi respondida e falsa (False) caso tenha ficado em branco. O operador AND pode ser usado para verificar se a prova foi concluída, uma vez que, para isso, todas as questões precisam ter sido feitas. Assim, para saber se alguém *não concluiu a prova*, basta negar essa condição:

```python
Prova_Nao_Concluida = not (Q1 and Q2 and Q3 and Q4)
```
\
Em Python, fica assim:

````python
# Nesse cenario, apenas as questões Q1, Q2 e Q3 foram feitas

Q1 = True
Q2 = True
Q3 = True
Q4 = False

prova_nao_concluida = not(Q1 and Q2 and Q3 and Q4)
print(prova_nao_concluida)
````

Esse exemplo mostra duas coisas novas:

* É possível negar uma expressão inteira;
* É possível misturar mais de um operador numa mesma expressão, assim como na matemática.

**Atenção:**

Como em equações comuns, existe uma hierarquia entre os operadores booleanos: `not`, `and` e `or`. Assim como na matemática, os parênteses alteram essa ordem. Veja mais sobre combinação de operadores na próxima seção.

### **Combinando expressões**

Como visto anteriormente, é possível combinar operações booleanas, formando equações mais complexas que podem ser úteis nos mais diversos tipos de problema.

Seguindo no exemplo do professor que dá aos seus alunos possibilidades de tirar nota máxima, considere um sistema de pontuação um pouco mais complexo:

```
QUESTÕES DA PROVA

CONJUNTO A (2 PONTOS):
QUESTÃO 1
QUESTÃO 2

CONJUNTO B (3 PONTOS):
QUESTÃO 3

CONJUNTO C (5 PONTOS):
QUESTÃO 4
QUESTÃO 5
QUESTÃO 6
```

Perceba que a lista está dividida em conjuntos e que cada conjunto tem sua pontuação.

```
CONJUNTO A (2 PONTOS)
CONJUNTO B (3 PONTOS)
CONJUNTO C (5 PONTOS)
```

Agora, considere a seguinte regra: o aluno deve resolver *pelo menos uma questão de cada conjunto* para obter a nota máxima na prova. O aluno pode, por exemplo, resolver as questões 1, 3 e 6, uma vez que cada uma pertence a um dos conjuntos apresentados. Dessa forma, o importante é *não deixar nenhum conjunto totalmente em branco*! Resolver mais exercícios do que o necessário não é um problema, mas também não resultará em um aumento da nota.

Assim, a expressão que modela esse cenário é:

```python
nota_10 = (Q1 or Q2) and Q3 and (Q4 or Q5
or Q6)
```

Use essa expressão e modifique as variáveis, onde `True` significa que a questão foi respondida, e `False` indica que a questão ficou em branco. Considere o caso onde as questões 1, 3 e 6 foram feitas:

```python
nota_10 = (True or False) and True and (False or False or True), #Isso equivale a:
nota_10 = True and True and True, #Isso equivale a:
nota_10 = True
```

De forma mais robusta, o seguinte código em Python modela a situação descrita:

````python
# Quais as diferentes formas de fazer com que o valor de `nota_10` seja 'True'?
Q1 = True
Q2 = True
Q3 = True
Q4 = False
Q5 = False
Q6 = True

nota_10 = (Q1 or Q2) and Q3 and (Q4 or Q5 or Q6)

print(nota_10)
````

# **O que pode dar errado?**

### **Operador `==`**

Para diferenciar um pouco os dois operadores, entenda o operador `=` como "recebe", e `==` como "igual a".

````python
numero_qualquer = 3.1415 # em linguagem natural: "numero_qualquer recebe 3,1415"
12 == 1 # em linguagem natural: "12 é igual a 1?"
````

### **Operadores `>=` e `<=`**

Um erro comum é inverter a ordem desses dois caracteres na escrita desse operador:

````python
12 >= 2 # código válido

12 => 2 # código inválido
````
