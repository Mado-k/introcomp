# **Operador elif**

Agora que você entendeu como os operadores if e else funcionam, é hora de analisar mais um membro da família dos operadores condicionais: o `elif`.

Observe o seguinte exemplo:

```python
if 'Primeira condição':
    'Alguma ação'

if 'Segunda condição':
    'Alguma ação'
```


Perceba que, nesta situação, os dois ifs são independentes - isto é, a primeira condição ser atendida não impede que a outra também seja! Agora, imagine: e se você quiser que uma **segunda condição** só seja testada se a primeira não for satisfeita? Para isso, usamos o operador `elif`.

O `elif` é uma abreviação do termo `else if` (se você já programou em linguagens como C, talvez conheça dessa forma), e significa
"caso contrário, se". Ele é útil para checar uma nova condição, quando a primeira (if) não for verdadeira. Com o uso do elif, seu código fica mais organizado e fácil de entender, principalmente quando você precisa lidar com várias possibilidades.


O esquema abaixo mostra um exemplo lúdico de utilização e sintaxe do elif:



```python
if 'Primeira condição':
    'Alguma ação'
elif 'Alguma outra condição':
    'Alguma outra ação'
elif 'Alguma outra condição, diferente de todas de cima':
    'Alguma outra ação, diferente das de cima'
else:
    'Alguma outra ação, caso nenhuma das condições sejam atendidas'  
```

Com o uso do elif, a segunda condição só será testada se a primeira for falsa. Da mesma forma, a terceira condição só será verificada se as duas anteriores forem falsas. Você pode escrever quantos elif quiser, mas apenas um if (primeira condição) e um else (caso nenhuma condição tenha sido atendida).

````python
# @title **Exemplo de Condições if, elif e else**
from IPython.display import display, Markdown, HTML
import ipywidgets as widgets


def condicao_exemplo(valor):
    codigo = f"""
valor = {valor}
if valor < 0:
    print("Valor negativo")
elif valor == 0:
    print("Valor é zero")
else:
    print("Valor positivo")
"""

    if valor < 0:
        resultado = "Valor negativo"
    elif valor == 0:
        resultado = "Valor é zero"
    else:
        resultado = "Valor positivo"

    html_style = """
    <style>
        .custom-box {
            background-color: #f2f2f2;
            color: #222222;
            border: 2px solid #cccccc;
            border-radius: 12px;
            padding: 15px;
            margin-top: 10px;
            font-family: monospace;
        }
        .output-box {
            font-size: 16px;
            font-weight: bold;
            color: #333333;
        }
        .widget-label {
            font-weight: bold;
        }
    </style>
    """

    html_code = f"""
    {html_style}
    <div class="custom-box">
        <pre>{codigo}</pre>
        <div class="output-box">Saída: {resultado}</div>
    </div>
    """
    display(HTML(html_code))


slider = widgets.IntSlider(
    value=0,
    min=-5,
    max=5,
    step=1,
    description='Valor:',
    continuous_update=False,
    style={'description_width': 'initial'}
)

interativo = widgets.interactive_output(condicao_exemplo, {'valor': slider})

display(HTML("<h2 style='font-weight: normal;'>Altere o valor abaixo para ver como o if, elif e else reagem:</h2>"))
display(Markdown("(Rode o código e deslize a bolinha!)"))
display(slider, interativo)
````

# **Aninhamento de Operadores**

Um último tema a ser abordado é o aninhamento de expressões `if`, `else` e `elif`:
Você pode escrever expressões condicionais dentro delas mesmas. Com isso, podemos criar condições dentro de condições, filtrando ainda mais nossos valores. Observe o exemplo abaixo:



```python
if condicaoInicial:
    if segundaCondicao:
        # Comandos e instruções
    elif terceiraCondicao:
        # Comandos e instruções caso segundaCondicao seja falsa
    else:
        # Comandos e instruções caso segundaCondicao e
        # terceiraCondicao sejam falsas
else:
  # Comandos e instruções caso condicaoInicial seja falsa
```




No exemplo, inicialmente, uma condição inicial é verificada. Caso ela seja verdadeira, uma segunda condição é verificada. Se a segunda condição for verdadeira, pode ser inserida alguma instrução, caso contrário, uma terceira condição é verificada, e assim por diante, até cair no caso onde todos as condicionais listadas no if correspondente sejam falsas.

Sinta-se à vontade para misturar e utilizar as estruturas condicionais de acordo com sua necessidade... mas cuidado para não se perder no código, pois podemos ter vários níveis de aninhamento!

Para tanto, é importante respeitar a `indentação` e os dois pontos que seguem essas expressões! Eles ditam quem está dentro de quem, dando sentido ao seu programa. No código a seguir, tente alterar os valores das variáveis e verifique os resultados. Adicione também novas condições `if/elif/else` aninhadas à condição inicial.

````python
# Vamos levar em conta três variáveis
num1 = 1
num2 = 5
num3 = 6

# Mais uma vez, mude as comparações e os valores e tente acrescentar casos de 'elif'
if num1 <= num2:
    if num1 > num3:
        print('Número 1 é menor ou igual ao Número 2 e maior que o Número 3')
    else:
        print('Número 1 é menor ou igual aos Números 2 e 3')
else:
    print('Número 1 é maior que o Número 2')
````

Perceba que você pode usar os operadores condicionais quando apropriado: um if dentro de outro representa um `and`!

```python
if primeira_condicao:
  if segunda_condicao:
    print('Primeira E segunda condições são verdadeiras!')
```

Uma forma equivalente de escrever o código acima:

```python
if primeira_condicao and segunda_condicao:
   print('Primeira E segunda condições são verdadeiras!')
```

Assuma que agora existem diferentes faixas de valor com descontos variáveis. <br>
* O plano **bronze** oferece agora 5% de desconto para compras de até R\$ 100,00 e 7,5% para compras acima desse valor. <br>
* O plano **prata** oferece 10% de desconto para compras de até R\$ 100,00 e 12,5% para compras acima desse valor. <br>
* Por fim, o plano **ouro** oferece 15% de desconto para compras de até R\$ 100,00, 17,5% para compras entre R\$ 100,00 e R\$ 200,00, e 20% para valores acima de R\$ 200,00.

**Observação:** Lembre-se de usar estruturas condicionais aninhadas para lidar com os diferentes casos.

````python
print('Bronze = 1 \nPrata = 2 \nOuro = 3')
plano = int(input())
valor = float(input())

if plano == 1:
  if valor <= 100:
    custo = valor * 0.05
  if valor > 100:
    custo = valor * 0.075

elif plano == 2 and valor >= 100:
  if valor <= 100:
    custo = valor * 0.10
  if valor > 100:
    custo = valor * 0.125

elif plano == 3 and valor >= 100:
  if valor <= 100:
    custo = valor * 0.15
  if valor > 100 and valor < 200:
    custo = valor * 0.175
  if valor >= 200:
    custo = valor * 0.20

print(valor - custo)
````
