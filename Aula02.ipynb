# **O que é um programa?**



Um programa é uma sequência de instruções que especificam como o computador deve realizar uma tarefa: resolver uma equação de segundo grau, encontrar e substituir uma palavra num texto, entre outros.  

Ele deve ser escrito em alguma **linguagem de programação**, que age como um idioma para facilitar a sua comunicação com o computador. Neste curso, você aprenderá a linguagem de programação **Python**.

\
Antes de mais nada, alguns conceitos básicos que farão parte do nosso vocabulário:

- **Operações Aritméticas**: obtenção de resultados com operações matemáticas (soma, subtração, divisão e etc);
- **Condicionais**: checar se certas condições são verdadeiras e realizar instruções apropriadas;
- **Repetições**: executar comandos de forma repetitiva;
- **Entrada**: receber informações, seja do teclado, de um arquivo, do mouse ou de um sensor de movimento numa câmera, por exemplo;
- **Saída**: enviar informações, seja num arquivo, num app de mensagens, entre outros.

# **Strings**





A manipulação de strings em programas é uma tarefa comum. Python oferece diversos métodos de manipulação, alguns exemplos são exibidos abaixo:

````python
# Tanto aspas simples ('string') quanto as aspas duplas ("string") fazem a mesma coisa!
frase_aspas_simples = 'Introcomp eh um curso muito legal'
print(frase_aspas_simples)

frase_aspas_duplas = "Introcomp eh um curso muito legal"
print(frase_aspas_duplas)
````

````python
# Apenas uma palavra também é considerado uma string!
palavra = "Introcomp"
print(palavra)
````

````python
# Note também que uma única letra também é uma string!
letra = "i"
print(letra)
````

````python
# O operador "in" informa se uma string existe dentro de uma outra.
# O resultado da operação retorna uma variável bool: True caso exista, False caso não exista.
frase = "Essa eh uma string de teste"
palavra = "teste"

verificacao = palavra in frase

print(verificacao)
````

````python
# Use o "+" para concatenar (juntar) strings!
frase = 'esta é uma frase'
frase_modificada = frase + " concatenada"
print(frase_modificada)
````

````python
# A função len informa o tamanho (número de caracteres) de uma string.
# Lembre-se que espaço, sinais de pontuação e qualquer outro
# símbolo individual (+, -, @, $, [, {, etc) também são caracteres:
string_longa = "Essa eh uma string de teste longa, qual o meu tamanho?"
tamanho = len(string_longa)
print(tamanho)
````

`in` e `len` nomeiam funções fornecidas nativamente pela linguagem Python. Por ora, entenda elas como "palavras especiais" que indicam uma operação específica, como uma "maquininha" que *recebe* algo (nesse caso, uma string), *processa* isso e *devolve* uma resposta. No momento, não se preocupe muito com detalhes técnicos.

Agora, veja algumas das principais funções para manipulacao de strings:

### **Função `lower`**



Transforma todos os caracteres que estão em maiúsculo para minúsculo ( A -> a )

````python
frase = "ISSO É UMA STRING"

frase_modificada = frase.lower()

print(frase)
print(frase_modificada)
````

### **Função `upper`**



Transforma todos os caracteres que estão em minúsculo para maiúsculo ( a -> A )

````python
frase = "está tudo pequeno"

frase_modificada = frase.upper()

print(frase)
print(frase_modificada)
````

Perceba que ao rodar os códigos, a variável original `frase` não foi alterada, isso acontece porque as funções `lower` e `upper`  não alteram a variável original, mas retornam uma **nova** string.

### **_f-strings_**

Uma maneira de inserir uma variável numa _string_. Por exemplo:

````python
nome = 'Fulano'
saudacao = f"Bom dia, {nome}!"

print(saudacao)
````

Perceba que o trecho onde existia um `{nome}` foi substituído pelo conteúdo da variável `nome`. Uma outra característica das _f-strings_ é a necessidade da letra `f` antes de abrir aspas.

Variáveis do tipo ponto flutuante (como a do exemplo) podem ser exibidas com um número menor de casas decimais, definido a partir do especificador **:.xf**, em que x é o número de casas decimais que deverão ser impressas:

````python
nome = 'Fulano'
imc = 20.2496

frase_poucos_decimais = f"{nome} possui o IMC igual a {imc:.2f} e está dentro do seu peso ideal!"
print("Usando f-string:")
print(frase_poucos_decimais)
````

### **A biblioteca 'math'**


Bibliotecas em Python são conjuntos prontos de códigos criados por outras pessoas para facilitar o desenvolvimento de programas. Elas funcionam como "caixas de ferramentas" que você pode usar para resolver problemas específicos, como fazer contas, desenhar gráficos, entre outros!\
No caso de operações matemáticas mais complexas, a biblioteca "math" se mostra muito útil. Basta digitar `import math` no começo do arquivo de código `.py`.

A seguir, alguns exemplos de funções da math:
1. `ceil(valor)` - arredonda `valor` para cima;
2. `floor(valor)` - arredonda `valor` para baixo;
3. `pow(valor1, valor2)` - retorna `valor1` elevado a `valor2`;
4. `sqrt(valor)` - retorna a raiz quadrada de `valor`;
5. `sin(valor), cos(valor), tan(valor)` - retorna, respectivamente, o seno, o cosseno e a tangente de `valor` (em radianos).
6. `factorial(valor)` - retorna o fatorial de `valor`;

Constantes, como o **Pi**, também são fornecidas por essa biblioteca. Para obter o valor de Pi, escreva `math.pi`

Caso você esteja curioso para ver o resto da documentação da biblioteca math, clique [aqui](https://docs.python.org/pt-br/3/library/math.html)


Alguns exemplos usando funções da biblioteca math:

````python
# Importar a biblioteca é necessario para utilizar suas ferramentas!
import math
````

````python
# Isso é uma potenciação
print(pow(5, 3))

# Se você preferir, pode fazer potenciações de outro jeito, sem importar a math
# ou seja, mudando a sintaxe mas mantendo a semântica
print(5**3)
````

````python
# Isso é uma radiciação
print(math.sqrt(121))

# Se você preferir, pode fazer radiciações de outro jeito, sem importar a math
# ou seja, mudando a sintaxe mas mantendo a semântica
print(121**0.5)
````

````python
# Executa o fatorial de um número
print(math.factorial(5))
````

````python
# Retorna o cosseno de um número (em rad)
print(math.cos(0))
````

````python
# Arredonda o número para cima
print(math.ceil(4.25))
````

# **Entradas**


Imagine um programa em Python que calcule a divisão entre dois números. Seria muito incoveniente ter que abrir o arquivo e modificar o valor das variáveis sempre que se desejasse fazer esse cálculo com outros valores, certo? É nessas horas que o comando `input()` entra em cena!\
Antes de tudo, perceba que sem a especificação do tipo, o `input()` naturalmente gera uma *string*:

````python
x = input()

# Aspas simples e duplas quase não têm diferença.
# A vantagem de misturar as duas é facilitar a escrita do código.
# Neste caso, as aspas simples foram usadas na frase
# para permitir imprimir aspas duplas dentro do texto sem gerar erro:
frase_aspas = f'A frase "{x}" não faz sentido!'
print(frase_aspas)

# Sem isso, seria necessário usar a barra invertida (\)
# para "escapar" as aspas:
frase_barra = f"A frase \"{x}\" não faz sentido!"
print(frase_barra)
````

Veja o tipo de x:

Para armazenar um número inteiro (**int**) ou ponto flutuante (**float**), utilize `int(input())` e `float(input())`, respectivamente.

````python
print("entre com um número inteiro: ")
a = int(input())
print("entre com um número quebrado: ")
b = float(input())
print(f"O número inteiro é {a};")
print(f"O número quebrado é {b};")
print(f"O resultado de {a}/{b} é {a/b:.2f}")
````
