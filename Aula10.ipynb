# **For**

Use o `for` quando tiver uma quantidade definida de itens para processar ou souber exatamente quantas repetições precisará fazer. Essas situações são comuns quando se trabalha com listas, sequências ou contagens específicas.

```python
# @title Exemplo interativo de for
from IPython.display import display, clear_output
import ipywidgets as widgets

# --- Configuração do Loop ---
sequencia = range(11)

# --- Estados ---
passo_atual = 0  # 0: Pegar o número, 1: Executar o print, 2: Verificar o próximo
indice = 0       # O índice atual na sequência (de 0 a 10)
saida_impressa = [] # Para simular a saída do terminal

# --- Widgets ---
codigo_label = widgets.HTML()
status_label = widgets.HTML()
acao_label = widgets.HTML()
conclusao_label = widgets.HTML()
terminal_label = widgets.HTML()

botao_proximo = widgets.Button(
    description="Próximo Passo",
    layout=widgets.Layout(width='160px', border_radius='12px'),
    style=widgets.ButtonStyle(button_color='#5A6BCF', font_weight='bold', text_color='#FFFFFF') # Azul suavizado
)

botao_reiniciar = widgets.Button(
    description="Reiniciar",
    layout=widgets.Layout(width='160px', border_radius='12px'),
    style=widgets.ButtonStyle(button_color='#7E5ECF', font_weight='bold', text_color='#FFFFFF') # Roxo suavizado
)

saida_principal = widgets.Output()

# --- Funções de Geração de HTML ---

def gerar_codigo_html():
    """Gera o HTML do bloco de código, destacando a linha atual."""
    linha_for_style = ""
    linha_print_style = ""

    if passo_atual == 0 or passo_atual == 2:
         linha_for_style = "background-color: #25573a; padding: 1px 4px; border-radius: 4px; color: #fff;"
    elif passo_atual == 1:
        linha_print_style = "background-color: #25573a; padding: 1px 4px; border-radius: 4px; color: #fff;"

    return f"""
    <div style='background-color: #2d2d2d; color: #d4d4d4; padding: 15px; border-radius: 8px; font-family: "Courier New", monospace; font-size: 16px; margin-bottom: 20px;'>
        <pre><code><span style='color: #c586c0;'>sequencia_0_a_10</span> = <span style='color: #4fc1ff;'>range</span>(<span style='color: #b5cea8;'>11</span>)

<span style='{linha_for_style}'><span style='color: #c586c0;'>for</span> numero <span style='color: #c586c0;'>in</span> sequencia_0_a_10:</span>
    <span style='{linha_print_style}'><span style='color: #4fc1ff;'>print</span>(numero)</span></code></pre>
    </div>
    """

def gerar_terminal_html():
    """Gera o HTML que simula a saída do 'print'."""
    conteudo = " ".join(map(str, saida_impressa)) if saida_impressa else "<i>(vazio)</i>"
    return f"""
    <div style='margin-top: 15px; border: 2px dashed #555; padding: 10px; border-radius: 8px; font-family: "Courier New", monospace; background-color: #1e1e1e; color: #fff; min-height: 24px;'>
        <div style='font-weight: bold; color: #888; margin-bottom: 5px;'>Saída do Terminal:</div>
        {conteudo}
    </div>
    """

def status_passo():
    """Retorna o texto descritivo do passo atual."""
    return f"Repetição {indice + 1} / Passo {passo_atual + 1}"

def acao_passo():
    """Retorna o texto da ação do passo atual."""
    if passo_atual == 0:
        numero_atual = sequencia[indice]
        return f"A variável `numero` recebe o valor <b>{numero_atual}</b>."
    elif passo_atual == 1:
        numero_atual = sequencia[indice]
        return f"Executando: `print({numero_atual})`"
    else: # passo_atual == 2
        if indice + 1 < len(sequencia):
            proximo_numero = sequencia[indice + 1]
            return f"Ainda há itens? ✅ Sim, o próximo é <b>{proximo_numero}</b>."
        else:
            return "Ainda há itens? ❌ Não, a sequência chegou ao fim."

def conclusao_passo():
    """Retorna o texto da conclusão (se houver)."""
    if passo_atual == 2:
        if indice + 1 < len(sequencia):
            return "O loop continuará. Clique em 'Próximo Passo'."
        else:
            return "<span style='color: #4CAF50; font-weight: bold;'>✅ Loop finalizado!</span>"
    return "<span style='visibility: hidden;'>.</span>"

def atualizar_interface():
    """Atualiza todos os widgets de texto da interface."""
    max_repeticoes = len(sequencia)
    if passo_atual == 2 and indice + 1 >= max_repeticoes:
        botao_proximo.disabled = True
    else:
        botao_proximo.disabled = False

    codigo_label.value = gerar_codigo_html()

    status_label.value = f"<div style='font-family: Courier New, monospace; font-size: 16px; text-align: center;'><b>{status_passo()}</b></div>"
    acao_label.value = f"<div style='font-family: Courier New, monospace; font-size: 16px; text-align: center; margin-top: 10px;'>{acao_passo()}</div>"
    conclusao_label.value = f"<div style='font-family: Courier New, monospace; font-size: 14px; text-align: center; margin-top: 15px;'><i>{conclusao_passo()}</i></div>"
    terminal_label.value = gerar_terminal_html()


# --- Funções de Lógica dos Botões ---

def avancar_passo(b):
    global passo_atual, indice, saida_impressa

    if passo_atual == 0:
        passo_atual = 1
    elif passo_atual == 1:
        numero_atual = sequencia[indice]
        saida_impressa.append(numero_atual)
        passo_atual = 2
    elif passo_atual == 2:
        if indice + 1 < len(sequencia):
            indice += 1
            passo_atual = 0

    atualizar_interface()

def reiniciar(b):
    global passo_atual, indice, saida_impressa
    passo_atual = 0
    indice = 0
    saida_impressa = []
    botao_proximo.disabled = False
    atualizar_interface()

# Conectar funções aos eventos de clique
botao_proximo.on_click(avancar_passo)
botao_reiniciar.on_click(reiniciar)


# --- Montagem e Exibição da Interface ---

box_principal = widgets.VBox([
    codigo_label,
    status_label,
    acao_label,
    conclusao_label,
    terminal_label
])

box_botoes = widgets.HBox(
    [botao_proximo, botao_reiniciar],
    layout=widgets.Layout(justify_content='center', margin='25px 0 0 0')
)

with saida_principal:
    clear_output()
    display(box_principal, box_botoes)

# Exibição inicial
reiniciar(None)
display(saida_principal)
```

## **Exercício 2**

Raphael está em um churrasco que terá **no máximo 10 rodadas** de comida. Em cada rodada, o prato servido será lido da entrada do teclado. O objetivo é contar e imprimir quantos pratos ele efetivamente comeu ao fim do dia, seguindo rigorosamente as seguintes regras:

* **Uma exceção:** Se for servido "peixe", ele pula essa rodada (não come e não conta no total).
* **Encerramento antecipado:** O churrasco sempre termina quando a sobremesa é servida. Portanto, se a entrada for "bolo", significa que as rodadas acabaram mais cedo; o programa deve computar o bolo como comido e interromper imediatamente a leitura das próximas rodadas.

**Exemplo de Entrada:**

picanha\
frango\
linguiça\
coração\
pão de alho\
salada\
bolo\
arroz\
farofa\
abacaxi

**Saída Correspondente:**

Raphael comeu 7 pratos!

---

**Exemplo de Entrada:**

picanha\
peixe\
frango\
pão de alho\
salada\
arroz\
linguiça\
farofa\
peixe\
abacaxi

**Saída Correspondente:**

Raphael comeu 8 pratos!



```python
pessoa = 'Raphael'
sobremesa = 'bolo'
pratos = []
prato = 0

for prato in range(10):
  comida = input()
  pratos.append(comida)

  if comida == sobremesa:
    prato += 1
    break
  elif comida != sobremesa:
    continue
  elif prato == 10:
    break

total = len(pratos)
print(f'{pessoa} comeu {total} pratos!')
```

## **Exercício 5**

Agora é hora de ir às compras. Como você é precavido, sabe que quando chegar vai precisar registrar o que comprou, então vai fazer um programa para agilizar isso.

Para cada elemento contido na lista **`itens`** (declarada no início do bloco de código, que está abaixo das entradas e saídas), o programa deve ler uma entrada que será `1` ou `0`.
* Se o item foi comprado (`1`), o programa deve pedir uma nova entrada com o preço dele (float) e somar ao total gasto.
* Se não foi comprado (`0`), o item deve ser guardado para ser exibido como restante no final.

Ao final, imprima o valor total gasto e a lista dos itens que ainda precisam ser comprados.

**Exemplo de Entrada:**

1\
20.00\
0\
1\
5.50\
1\
7.00

**Saída Correspondente:**

Total gasto: 32.50\
Itens restantes:\
feijão

---

**Exemplo de Entrada:**

0\
0\
1\
4.99\
0

**Saída Correspondente:**

Total gasto: 4.99\
Itens restantes:\
arroz\
feijão\
pão


```python
itens = ['arroz', 'feijão', 'leite', 'pão']
total = 0
lista_comprados = []

for i in itens:
  comprado = int(input())

  if comprado == 0:
    pass
  elif comprado == 1:
    preco = float(input())
    total += preco
    lista_comprados.append(i)

for i in itens:
  if i in lista_comprados:
    itens.remove(i)

print(f'Total gasto: {total}')
print(f'Itens restantes: ')
for item in itens:
  print(item)
```

## **Exercício 6**

A turma do Introcomp foi muito bem nas atividades e recebeu uma recompensa: todos os alunos receberão 1 ponto extra na nota da prova. Entretanto, os alunos que já tem nota 10 não deverão ter a nota incrementada. Para isso, seu programa deve receber como entrada as notas da turma separadas por vírgula, e então exibir a nova coleção de notas.

**Exemplo de entrada:**\
10,9,8,7,10,9,0,5

**Exemplo de saida:**

[10, 10, 9, 8, 10, 10, 1, 6]

```python
n = input()
n = n.split(',')

for i in range(len(n)):
  n[i] = int(n[i])

for i in range(len(n)):
  if n[i] < 10:
    n[i] += 1
  else:
    continue

print(n)
```

## **Exercício 5**

```python
itens = ['arroz', 'feijão', 'leite', 'pão']

total = 0.0
itens_restantes = []

for item in itens:
    flag = input(f"O item {item} foi comprado? ")
    if flag == "1":
        # se comprou, lê o preço e soma ao total
        preco = float(input("Digite o preço: "))
        total += preco
    else:
        # se não comprou (0), adiciona na lista de restantes
        itens_restantes.append(item)

print(f"Total gasto: {total:.2f}")
print("Itens restantes:")
for i in itens_restantes:
    print(i)
```
